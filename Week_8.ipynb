# Week 8 Demo Outline
## Question

Can machine learning predict Click-Through Rate (CTR) using historical search performance data?

## Method

- Used the FlyRank ML Internship Warehouse dataset.
- Aggregated daily search metrics to page level using DuckDB.
- Selected numerical search performance features.
- Trained Linear Regression and Random Forest regression models.
- Evaluated performance using MAE, MSE and R².

## One Chart

Feature Importance chart

## Honest Result

Linear Regression and Random Forest were able to identify patterns in CTR, but the low R² values indicate that the available features explain only a small portion of CTR variation.

## Recommendation

Future work should include richer behavioural and content-quality features to improve prediction accuracy and better identify optimisation opportunities.
